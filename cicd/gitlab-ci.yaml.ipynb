# `.gitlab-ci.yaml` 파일

- CI/CD는 `.gitlab-ci.yml` 이라는 이름의 yaml 형식파일로 정의
- `.gitlab-ci.yml` 파일은 하나의 레포에 하나만 존재


## yaml 형식 파일

- `.yaml` 또는 `.yaml` 확장자를 가지는 텍스트 파일
- `#` 기호로 시작하는 주석
- 노드가 트리 형태로 연결된 구조를 가짐
- 트리의 최상위 노드는 무조건 맵
- 하위 노드는 맵, 리스트, 문자열 중 하나
- 맵
  - 키(key, 이름)를 가지는 자식 노드가 있는 노드
  - 자식 노드의 순서는 의미 없음
  - `키문자열: ` 로 시작함 
   - 키문자열과 콜론(``) 기호 사이에는 공백이 없어야 하고 콜론 기호 뒤에는 하나의 공백이 있어야 함
   - 키문자열은 공백이나 특수문자 포함 가능
- 리스트
  - 이름이 없이 순서로만 정의되는 자식 노드가 있는 노드
  - 자식 노드의 순서가 의미를 가짐
  - 대시(`-`) 기호로 시작
    - 대시(`-`) 기호 뒤에는 하나의 공백이 있어야 함
- 문자열
  - 자식 노드 없이 문자열로 정의되는 노드
  - 콜론이나 대시 기호 뒤에 바로 문자열이 옴 
  - 모든 말단 노드는 문자열 노드



### 예시 1

- 최상위 레벨은 맵
- 최상위 맵의 키는 다음 3개
  - key-1-1
  - key-1-2
  - key-1-3
- key-1-1 키의 값은 문자열
- key-1-2 키의 값은 리스트
- key-1-3 키의 값은 맵
- key-1-3 키 값 맵의 키는 다음 2개
  - key-1-3-1
  - key-1-3-2
  - key-1-3-3
- key-1-3-1 키의 값은 문자열
- key-1-3-2 키의 값은 리스트
- key-1-3-3 키의 값은 맵

  

```yaml
# 최상위 레벨 노드는 맵 노드
key-1-1: value 1 1                 # 문자열 노드
  
key-1-2:                           # 리스트 노드
  - list 1 2 1                     #   문자열 노드
  - list 1 2 2                     #   문자열 노드
  - list 1 2 3                     #   문자열 노드

key-1-3:                           # 맵 노드
  key-1-3-1: value 1 3 1           #   문자열 노드
  key-1-3-2:                       #   리스트 노드
    - list 1 3 2 1                 #     문자열 노드
    - list 1 3 2 2                 #     문자열 노드
  key-1-3-3:                       #   맵 노드
    key-1-3-3-1: value 1 2 3 1     #     문자열 노드 
    key-1-3-3-2: value 1 2 3 2     #     문자열 노드
```

## `.gitlab-ci.yml` 파일

- 파이프라인과 그 내부의 job을 정의하는 yaml 형식 파일
- `.gitlab-ci.yml` 파일의 키에는 2종류가 있음
  - 미리 GitLab에서 정의하여 사용하는 CI/CD 키워드 문자열
  - 사용자가 정의한 job 이름 문자열
- job 이름 문자열과 일부 CI/CD 키워드는 반드시 최상위 레벨 키
- job은 항상 맵 노드
- CI/CD 키워드는 종류별로 타입이 다름

### 예시 파일

- 4개의 최상위 키 `build-job`, `test-job1`, `test-job2`, `deploy-prod` 는 모두 사용자가 정의한 job의 이름
- 하위의 `stage`, `script` 는 미리 정의된 CI/CD 키워드
  - stage
    - job의 실행 순서를 나타내는 문자열
    - 순서를 나타내는 `build`, `test`, `deploy` 중 하나의 문자열 값을 가짐
    - 사용자가 스테이지를 별도로 정의한 경우에는 다른 값을 가질 수도 있음
  - script
    - job이 실제로 실행하는 bash 명령을 나타내는 문자열의 리스트


```yaml
build-job:
  stage: build
  script:
    - echo "Hello, $GITLAB_USER_LOGIN!"

test-job1:
  stage: test
  script:
    - echo "This job tests something"

test-job2:
  stage: test
  script:
    - echo "This job tests something, but takes more time than test-job1."
    - echo "After the echo commands complete, it runs the sleep command for 20 seconds"
    - echo "which simulates a test that runs 20 seconds longer than test-job1"
    - sleep 20

deploy-prod:
  stage: deploy
  script:
    - echo "This job deploys something from the $CI_COMMIT_BRANCH branch."
    
```

## CI/CD 키워드

- GitLab에서 미리 정의해 놓은 CI/CD 설정용 키 값 문자열
- 키워드와 같은 job 이름은 사용자가 정의할 수 없음
- 약 40개 정도의 키워드가 정의되어 있음
- 전체 키워드 목록 (알파벳 순)
    - 전역 키워드
      - `default`	
      - `include`	
      - `stages`	
      - `variables`
      - `workflow`
    - 헤더 키워드
      - `spec`	
    - Job 키워드
      - `after_script`
      - `allow_failure`
      - `artifacts`	
      - `before_script`	
      - `cache`	
      - `coverage`	
      - `dast_configuration`	
      - `dependencies`
      - `environment`	
      - `extends`	
      - `identity`	
      - `image`	
      - `inherit`	
      - `interruptible`
      - `manual_confirmation`
      - `needs`	
      - `pages`	
      - `parallel`	
      - `release`	
      - `resource_group`	
      - `retry`	
      - `rules`	
      - `script`	
      - `run`	
      - `secrets`	
      - `services`	
      - `stage`	
      - `start_in`	
      - `tags`	    
      - `timeout`	
      - `trigger`	
      - `variables`	
      - `when`
