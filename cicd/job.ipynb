# job

## job 이름

- 일반적인 yaml 규칙과 달리 job 이름을 정의하는 키 문자열은 공백, 특수문자 등을 포함 가능 

## hidden job

- 마침표(`.`)로 시작하는 job은 실제로 실행되지 않는 hidden job이다.
- 다만 이 hidden job이 anchor로 정의되어 다른 job에서 사용될 수는 있다. 

## anchor, alias, map merge

- 반복되는 중복 설정을 줄이기 위한 yaml 자체 기능
- anchor 기능으로 맵이나 리스트에 이름을 정의 
- 정의된 anchor 노드를 
  - alias 기능으로 다른 자식노드로 설정
  - map merge 기능으로 다른 맵 요소 정의

```yaml
.dummy1: &anchor1
  - value1
  - value2

job1: *anchor1

job2:
  <<: *anchor1
  - value3
```
