# pipeline

- 파이프라인은 하나의 레포지토리가 가지는 CI/CD 실행 단위
- `.gitlab-ci.yml` 이라는 이름의 yaml 형식파일로 정의
- 기본적으로는 하나의 레포지토리는 하나의 파이프라인을 가질 수 있음
  - include, local 등을 사용한 고급 설정을 통해 복수의 파이프라인 구성도 가능
- 하나의 파이프라인은 여러개의 job 으로 구성
- job은 순차적 또는 병렬적으로 실행 가능


## yaml 형식 파일

- `.yaml` 또는 `.yaml` 확장자를 가지는 텍스트 파일
- 노드가 트리 형태로 연결된 구조를 가짐
- 트리의 최상위 노드는 무조건 맵
- 하위 노드는 맵, 리스트, 문자열 중 하나
- 맵
  - 키(key, 이름)를 가지는 자식 노드가 있는 노드
  - 자식 노드의 순서는 의미 없음
  - `키문자열: ` 로 시작함 
   - 키문자열과 콜론(`:`) 기호 사이에는 공백이 없어야 하고 콜론 기호 뒤에는 하나의 공백이 있어야 함
   - 키문자열은 공백이나 특수문자 포함 가능
- 리스트
  - 이름이 없이 순서로만 정의되는 자식 노드가 있는 노드
  - 자식 노드의 순서가 의미를 가짐
  - 대시(`-`) 기호로 시작
    - 대시(`-`) 기호 뒤에는 하나의 공백이 있어야 함
- 문자열
  - 자식 노드 없이 문자열로 정의되는 노드
  - 모든 말단 노드는 문자열 노드



### 예시

- 최상위 레벨은 맵
- 최상위 맵의 키는 다음 3개
  - key-1-1
  - key-1-2
  - key-1-3
- key-1-1 키의 값은 문자열
- key-1-2 키의 값은 리스트
- key-1-3 키의 값은 맵
- key-1-3 키 값 맵의 키는 다음 2개
  - key-1-3-1
  - key-1-3-2
  - key-1-3-3
- key-1-3-1 키의 값은 문자열
- key-1-3-2 키의 값은 리스트
- key-1-3-3 키의 값은 맵

  

```yaml

key-1-1: value 1 1
  
key-1-2:
  - list 1 2 1
  - list 1 2 2
  - list 1 2 3

key-1-3:
  key-1-3-1: value 1 3 1
  key-1-3-2: 
    - list 1 3 2 1
    - list 1 3 2 2
  key-1-3-3:
    key-1-3-3-1: value 1 2 3 1
    key-1-3-3-2: value 1 2 3 2
```

## `.gitlab-ci.yml` 파일

- 파이프라인과 그 내부의 job을 정의하는 yaml 형식 파일
- `.gitlab-ci.yml` 파일의 키에는 2종류가 있음
  - 미리 GitLab에서 정의하여 사용하는 CI/CD 키워드 문자열
  - 사용자가 정의한 job 이름 문자열
- 키에 대응하는 값의 형태는 키에 따라 다름
  - CI/CD 키워드는 키워드마다 미리 정해진 형태의 값을 가져야 함

### 예시 파일

- 4개의 최상위 키 `build-job`, `test-job1`, `test-job2`, `deploy-prod` 는 모두 사용자가 정의한 job의 이름
- 하위의 `stage`, `script`, `environment` 는 미리 정의된 CI/CD 키워드
- `stage` 키는 `build`, `test`, `deploy` 중 하나인 단일 문자열 값을 가짐
  - 사용자가 스테이지를 별도로 정의한 경우에는 다른 값을 가질 수도 있음
- `script` 키는 bash 명령어 문자열의 리스트 값을 가짐


```yaml
build-job:
  stage: build
  script:
    - echo "Hello, $GITLAB_USER_LOGIN!"

test-job1:
  stage: test
  script:
    - echo "This job tests something"

test-job2:
  stage: test
  script:
    - echo "This job tests something, but takes more time than test-job1."
    - echo "After the echo commands complete, it runs the sleep command for 20 seconds"
    - echo "which simulates a test that runs 20 seconds longer than test-job1"
    - sleep 20

deploy-prod:
  stage: deploy
  script:
    - echo "This job deploys something from the $CI_COMMIT_BRANCH branch."
  environment: production
```
