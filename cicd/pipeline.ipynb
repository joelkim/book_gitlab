# pipeline

- 파이프라인은 하나의 레포지토리가 가지는 CI/CD 실행 단위
- `.gitlab-ci.yml` 이라는 이름의 yaml 형식파일로 정의
- 기본적으로는 하나의 레포지토리는 하나의 파이프라인을 가질 수 있음
  - include, local 등을 사용한 고급 설정을 통해 복수의 파이프라인 구성도 가능
- 하나의 파이프라인은 여러개의 job 으로 구성
- job은 순차적 또는 병렬적으로 실행 가능


## 스테이지

- job을 순차적으로 실행하기 위한 일종의 태그값
- job 노드의 `stage` 키로 정의
- 사용자가 별도로 정의하지 않는 경우 `.pre`, `build`, `test`, `deploy`, `.post` 5개 값을 가질 수 있으며 이 순서대로 실행된다.
- 동일한 스테이지값을 가진 job은 병렬로 실행된다.
- 앞 스테이지의 모든 job이 성공적으로 종료되지 않으면 다음 스테이지는 실행되지 않음
